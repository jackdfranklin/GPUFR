```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read data from the file
file_path = "times.csv"
size = []
firefly = []
gpufr = []

with open(file_path, 'r') as file:
    for idx, line in enumerate(file):
        parts = line.strip().split()
        if len(parts) == 3 and idx>0:
            size.append(int(parts[0]))
            firefly.append(float(parts[1]))
            gpufr.append(float(parts[2]))

size = np.array(size)
firefly = np.array(firefly)
gpufr = np.array(gpufr)
```

```python
plt.loglog(size, firefly, label='FireFly', marker='o')
plt.loglog(size, gpufr, label='GPUFR', marker='o')

plt.legend()
plt.xlabel("poly size")
plt.ylabel("execution time [s]")
```
